# Problem 1
Write a function called `average_score` that calculates the average of an arbitrary number of scores.

**Sample output**
```
average_score(85, 90, 78) -> 84.333...
average_score(100, 95) -> 97.5
average_score() -> 0
```

````python
# TODO write the average_score function
def average_score(*scores) -> float:
    if not scores:
        return 0
    return sum(scores) / len(scores)


# Example usage:
print(average_score(85, 90, 78))  # Output: 84.333...
print(average_score(100, 95))  # Output: 97.5
print(average_score())  # Output: 0 (no scores given)
print(f"Average Score for Arite : {average_score(90, 60, 90):.2f}")
print(f"Average Score for Denise: {average_score(75, 55, 80):.2f}")
print(f"Average Score for Karina: {average_score(80, 50, 85):.2f}")
print(f"Average Score for Doreen: {average_score(99, 66, 96):.2f}")
print(f"Average Score for Carmen: {average_score(70, 50, 70):.2f}")
print(f"Average Score for Anke M: {average_score(85, 65, 85):.2f}")
````

# Problem 2
Write a function `describe_pet` that accepts a pet’s name and type as required arguments, but can also take any number of keyword details.

**Sample output**
```
Pet name: Whiskers
Animal type: cat
Details:
  color: gray
  age: 3
  favorite_food: tuna
```

````python
# TODO write the describe_pet function
def describe_pet(name: str, animal_type: str, **attributes) -> None:
    print(f"Pet Name: {name}")
    print(f"Animal Type: {animal_type}")
    if attributes:
        print("Additional Attributes:")
    for attr, value in attributes.items():
        print(f"  {attr.replace('_', ' ').title()}: {value}")


# Example usage:
describe_pet("Whiskers", "cat", color="gray", age=3, favorite_food="tuna")
print("-" * 40)
describe_pet(
    "Kater Karlo",
    "Katze",
    Farbe="schwarz weiß",
    Alter=4,
    lieblings_futter="Gourmet Perle",
)
print("-" * 40)
describe_pet(
    "Arite Zastrow",
    "Fick Stück",
    rasse="deutsch",
    alter=39,
    gewicht="60kg",
    Titten="85 C",
    muschi="sehr haarig",
    lieblings_stellung="Missionar geklammert",
)
````

# Problem 3
Write a function `make_sandwich` that accepts any number of ingredients and any number of optional preferences.

**Sample output**
```
Making a sandwich with:
  turkey
  lettuce
  tomato
Options:
  sauce: mayo
  toasted: True
```

````python
# TODO Write the make_sandwich function
def make_sandwich(*ingredients, **options) -> None:
    print("Making a sandwich with:")
    for ingredient in ingredients:
        print(f"  {ingredient}")

    if options:
        print("Additional Attributes:")
    for key, value in options.items():
        print(f"  {key.replace('_', ' ').title()}: {value}")


# Example usage:
make_sandwich("turkey", "lettuce", "tomato", sauce="mayo", toasted=True)
print("-" * 40)
make_sandwich(
    "Arite Zastrow",
    "René Kallenbach",
    "Denise Born",
    rasiert="ja, alle blitzblank",
    Kondom="Nein, auf keinen Fall",
    gleitmittel="Flutschi",
)
print("-" * 40)
make_sandwich(
    "Enrico S.",
    "Karina Hemmo",
    "Daniel B.",
    Stellung="Hündchen - oral & vaginal",
    rasiert="Karina, blitzblank",
)
````
