### Problem 1
Its important to assign return values to variables if you'd like to use them later <br> in your program.

What will be the output of the following code?

```python
def greet(name) -> str:
    return "Hello " + name + "!"


greeting = greet("Alice")
greet("Bob")
print(greeting)
print(greet("Arite Zastrow"))
print(greet("Denise 'bald pussy' Born"))
print(greet("Karina Hemmo, die Sportliche"))
print(greet("Doreen Barthel mit den Mega Möpsen und dem haarigen Monster"))
print(
    greet("Carmen Hanko - kein Arsch und kein Tittchen, sieht aus wie Schneewittchen")
)
```

### Problem 2

The return value of one function call can be used as an argument to another <br> function call.

What will this code output?


```python
def add(a, b):
    return a + b


result = add(5, 10)
print(result)
print(add(result, 20))
print(add(add(1, 2), add(3, 4)))
print(add(1.1, 2.2))
print(add(add("Hallo ", "Arite"), "!"))
```

### Problem 3

Python allows us to use multiple return statements inside a function.

What will this code output?


```python
def compute(a, b):
    result = a * b
    print(f"{a} * {b} = {result}")
    if result > 100:
        return result / 10
    if result > 10:
        return result - 3
    return result + 4


print(compute(3, 2))
print(compute(5, 3))
print(compute(3, 33))
print(compute(3.33, 33.33))
print(compute(add(1.1, 2.2), add(3.3, 4.4)))
```
