# File Handling - Exercises

## Problem 1

Write a program that creates a text file called `fruits.txt` with three fruits — one per line: `apple`, `banana`, and `cherry`. Then read the file back and print its contents.

````python
from typing import List
fruits: List[str] = ['apple', 'banana', 'cherry', 'strawberry']
with open('fruits.txt', 'w', encoding='utf-8') as f:
    for fruit in fruits:
        f.write(fruit + '\n')
print('Fruits have been written to fruits.txt')        
with open('fruits.txt', 'r', encoding='utf-8') as f:
    contents: str = f.read()
print('Content of fruits.txt:')    
print(contents)
````

## Problem 2

You are given the following string of city names:

```python
cities = "Tokyo\nLondon\nParis\nBerlin"
```

Write the cities to a file called `cities.txt`. Then read the file back, split the contents into a list of individual cities, sort them alphabetically, and print each city with a leading dash.

**Sample output:**
```
- Berlin
- London
- Paris
- Tokyo
```

````python
# TODO: Write cities to a file, read them back, sort, and print with dashes
cities: str = "Tokyo\nLondon\nParis\nPeiz\nPeking\nBerlin\nDakar"
with open('cities.txt', 'w', encoding='utf-8') as f:
    f.write(cities)

print('Cities have been written to cities.txt')

with open('cities.txt', 'r', encoding='utf-8') as f:
    city_list: List[str] = f.read().splitlines()

city_list.sort()
print('Sorted cities:')
# print(' - ' + '\n - '.join(city_list))    
for city in city_list:
    print(' - ' + city)
````

## Problem 3

Using context managers (`with` statements), create a file called `journal.txt` and write the entry `"Day 1: Started learning Python"` to it. Then, using another context manager, append two more entries on separate lines: `"Day 2: Practiced file handling"` and `"Day 3: Built my first project"`. Finally, read and print the complete journal.

**Sample output:**
```
Day 1: Started learning Python
Day 2: Practiced file handling
Day 3: Built my first project
```

````python
# TODO: Use context managers to write, append, and read journal entries
with open('journal.txt', 'w', encoding='utf-8') as f:
    f.write('Day 1: Started learning Python.\n')
with open('journal.txt', 'a', encoding='utf-8') as f:
    f.write('Day 2: Practiced file handling.\n')
    f.write('Day 3: Build my first project.')
with open('journal.txt', 'r', encoding='utf-8') as f:
    journal_contents: str = f.read()
print('Journal entries:')
print(journal_contents)
````
